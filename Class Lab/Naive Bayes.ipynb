```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import StratifiedShuffleSplit
from sklearn.naive_bayes import GaussianNB
```

```python
data = {'height':[6,5.92,5.58,5.92,5.00,5.50,5.42,5.75],
     'weight':[180,190,170,165,100,150,130,150],
     'foot_size':[12,11,12,10,6,8,7,9],
     'gender':['male','male','male','male','female','female','female','female']}
```

```python
df_data = pd.DataFrame(data= data)
df_data
```

```python
le = LabelEncoder()

df_data['gender'] = le.fit_transform(df_data['gender'])
```

```python
display(df_data)
```

```python
corr_matrix = df_data.corr(numeric_only= True)
corr_matrix
corr_matrix['gender'].sort_values(ascending = False)
```

```python
splitter = StratifiedShuffleSplit(n_splits = 10, test_size = 0.2, random_state = 42)
strat_splits = []
for train_index, test_index in splitter.split(df_data, df_data["gender"]):
    strat_train_set_n = df_data.iloc[train_index]
    strat_test_set_n = df_data.iloc[test_index]
    strat_splits.append([strat_train_set_n, strat_test_set_n])
```

```python
train_set = strat_train_set_n.copy()
attr = train_set.columns
train_set_x = train_set[attr[:len(attr)-1]]
train_set_y = train_set[attr[len(attr)-1:]]
```

```python
clf = GaussianNB()
clf.fit(train_set_x, train_set_y)
clf.score(train_set_x, train_set_y)
```

```python
test_set = strat_test_set_n.copy()
attr = test_set.columns
test_set_x = test_set[attr[:len(attr)-1]]
test_set_y = test_set[attr[len(attr)-1:]]
```

```python
result = clf.predict(test_set_x)
print(result)
print(test_set_y)
```

```python
matches = [predicted == y for predicted, y in zip(result, test_set_y.values)]
pd.DataFrame(matches).value_counts()
```

```python
# Predict New Number
x = pd.DataFrame([[5,100,6]],columns=['height','weight','foot_size'])
predicted = clf.predict(x)
print(f"Predicted Gender: {'Male' if predicted else 'Female'}")
```
